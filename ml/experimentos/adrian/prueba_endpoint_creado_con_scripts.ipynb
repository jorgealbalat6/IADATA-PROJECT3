```python
from google.cloud import aiplatform

# 1. Le decimos a Python dónde está tu ventanilla (tu Endpoint)
endpoint_name = "projects/926808849442/locations/europe-west1/endpoints/822058664599748608"

# 2. Conectamos con el Endpoint
endpoint = aiplatform.Endpoint(endpoint_name)

# 3. Preparamos un "cliente" ficticio de Airbnb (Sustituye estos ceros por valores reales de tu tabla)
# Importante: El XGBoost espera una lista de listas con los números en el mismo orden que en el entrenamiento
datos_de_prueba = [
    [0.0, 5, 2, 0, 15, 0, 120.5, 2, 45, 4.8, 1, 1, 22.5, 0.0, 1, 0, 50000] 
]

# 4. Enviamos los datos a la ventanilla y esperamos la respuesta
try:
    print("Enviando datos al modelo en Google Cloud...")
    respuesta = endpoint.predict(instances=datos_de_prueba)
    
    print("\n¡Éxito! El modelo ha respondido:")
    print(f"Predicción (is_occupied): {respuesta.predictions[0]}")
    
except Exception as e:
    print(f"\nHa ocurrido un error al preguntar al modelo: {e}")


#FUNCIONAAA!! Despues de haber pasado el testeo, he borrado el endpoint.
```
